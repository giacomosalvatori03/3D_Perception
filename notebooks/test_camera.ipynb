```python
import sys

sys.path.insert(0, "/content/3D_Perception")
from src import CameraDetector

# Inizializza il rilevatore 2D con pesi KITTI
detector_2d = CameraDetector(
    repo_id="dronefreak/kitti-yolov8m",
    filename="model.pt",
    conf_thresh=0.25,
    device="cuda",
)

# Test dell'inferenza su un frame di validazione
sample_img_path = (
    "/content/drive/MyDrive/3D_Perception/data/kitti_validation/image_2/000008.png"
)
detections_2d = detector_2d.detect(sample_img_path)

print(f"\n Trovate {len(detections_2d)} predizioni 2D nell'immagine:")
for det in detections_2d:
    bbox = [round(v, 1) for v in det["bbox_2d"]]
    print(
        f"  - [{det['type']}] Confidenza: {det['score']:.2f} | BBox 2D: {bbox}"
    )
```

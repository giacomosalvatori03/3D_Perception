```python
import os
import glob
import numpy as np

exp_dir = "/content/drive/MyDrive/3D_Perception/experiments/baseline_100perc"
gt_dir = os.path.join(exp_dir, "gt_labels")
pred_dir = os.path.join(exp_dir, "pred_labels")

gt_files = sorted(glob.glob(os.path.join(gt_dir, "*.txt")))

if not gt_files:
    print("❌ Nessun file GT trovato nella cartella!")
else:
    sample_file = os.path.basename(gt_files[0])
    print(f"🔍 ANALISI DIAGNOSTICA SUL FRAME: {sample_file}\n")

    # 1. Lettura Ground Truth
    gt_objects = []
    with open(os.path.join(gt_dir, sample_file), "r") as f:
        for line in f:
            p = line.strip().split()
            if len(p) >= 14:
                gt_objects.append({
                    'type': p[0],
                    'bbox2d': [float(x) for x in p[4:8]],
                    'dim': [float(x) for x in p[8:11]],   # h, w, l
                    'loc': [float(x) for x in p[11:14]],  # x, y, z
                    'ry': float(p[14])
                })

    # 2. Lettura Predizioni
    pred_objects = []
    pred_path = os.path.join(pred_dir, sample_file)
    if os.path.exists(pred_path):
        with open(pred_path, "r") as f:
            for line in f:
                p = line.strip().split()
                if len(p) >= 15:
                    pred_objects.append({
                        'type': p[0],
                        'bbox2d': [float(x) for x in p[4:8]],
                        'dim': [float(x) for x in p[8:11]],   # h, w, l
                        'loc': [float(x) for x in p[11:14]],  # x, y, z
                        'ry': float(p[14]),
                        'score': float(p[15]) if len(p) >= 16 else 1.0
                    })

    print(f"📦 Totale oggetti nel frame -> GT: {len(gt_objects)} | Predizioni: {len(pred_objects)}\n")

    # Stampa i primi oggetti per confronto visivo
    print("--- PRIMO OGGETTO GROUND TRUTH ---")
    if gt_objects:
        gt0 = gt_objects[0]
        print(f"  Tipo: {gt0['type']}")
        print(f"  Posizione 3D (x, y, z): {gt0['loc']}")
        print(f"  Dimensioni (h, w, l):  {gt0['dim']}")
        print(f"  Box 2D (xmin, ymin, xmax, ymax): {gt0['bbox2d']}")
    else:
        print("  [Nessun oggetto GT]")

    print("\n--- PRIMA PREDIZIONE ---")
    if pred_objects:
        p0 = pred_objects[0]
        print(f"  Tipo: {p0['type']}")
        print(f"  Posizione 3D (x, y, z): {p0['loc']}")
        print(f"  Dimensioni (h, w, l):  {p0['dim']}")
        print(f"  Box 2D (xmin, ymin, xmax, ymax): {p0['bbox2d']}")
        print(f"  Confidence Score: {p0['score']}")
    else:
        print("  [Nessuna predizione]")

    # 3. Calcolo Distanza Euclidea tra i centri 3D delle 'Car'
    print("\n--- MATRICI DI DISTANZA CENTRI 3D (Classe 'Car') ---")
    gt_cars = [g for g in gt_objects if g['type'] == 'Car']
    pred_cars = [p for p in pred_objects if p['type'] == 'Car']

    if gt_cars and pred_cars:
        min_dist = float('inf')
        for g in gt_cars:
            for p in pred_cars:
                dist = np.linalg.norm(np.array(g['loc']) - np.array(p['loc']))
                if dist < min_dist:
                    min_dist = dist
        print(f"🎯 Minima distanza 3D trovata tra una GT e una Predizione: {min_dist:.2f} metri")
        if min_dist > 3.0:
            print("⚠️ ATTENZIONE: La distanza tra centri è troppo grande (> 3m)! I box non si sovrappongono.")
    else:
        print("⚠️ Nessuna corrispondenza di classe 'Car' trovata tra GT e Predizioni.")
```

```python
import os
import glob
import numpy as np
from evaluate import load_kitti_txt, compute_iou_3d

exp_dir = "/content/drive/MyDrive/3D_Perception/experiments/baseline_100perc"
gt_dir = os.path.join(exp_dir, "gt_labels")
pred_dir = os.path.join(exp_dir, "pred_labels")

gt_files = sorted(glob.glob(os.path.join(gt_dir, "*.txt")))

total_gt_cars = 0
total_pred_cars = 0
max_iou_standard = 0.0
max_iou_adjusted_y = 0.0

for gt_path in gt_files:
    fname = os.path.basename(gt_path)
    pred_path = os.path.join(pred_dir, fname)

    gt = load_kitti_txt(gt_path, is_pred=False)
    pred = load_kitti_txt(pred_path, is_pred=True)

    if gt is None or pred is None:
        continue

    # Conteggio auto
    gt_car_mask = (gt['name'] == 'Car')
    pred_car_mask = (pred['name'] == 'Car')

    total_gt_cars += np.sum(gt_car_mask)
    total_pred_cars += np.sum(pred_car_mask)

    gt_locs = gt['location'][gt_car_mask]
    gt_dims = gt['dimensions'][gt_car_mask]
    gt_rys = gt['rotation_y'][gt_car_mask]

    pred_locs = pred['location'][pred_car_mask]
    pred_dims = pred['dimensions'][pred_car_mask]
    pred_rys = pred['rotation_y'][pred_car_mask]

    # Confronto IoU 3D su tutte le coppie di Car nel frame
    for i in range(len(gt_locs)):
        for j in range(len(pred_locs)):
            # 1. IoU Standard
            iou_std = compute_iou_3d(gt_locs[i], gt_dims[i], gt_rys[i], pred_locs[j], pred_dims[j], pred_rys[j])
            if iou_std > max_iou_standard:
                max_iou_standard = iou_std

            # 2. IoU con aggiustamento centro Y (pred_y - h/2)
            adj_pred_loc = pred_locs[j].copy()
            adj_pred_loc[1] = adj_pred_loc[1] + (pred_dims[j][0] / 2.0)  # Traslazione base/centro
            iou_adj = compute_iou_3d(gt_locs[i], gt_dims[i], gt_rys[i], adj_pred_loc, pred_dims[j], pred_rys[j])
            if iou_adj > max_iou_adjusted_y:
                max_iou_adjusted_y = iou_adj

print("=" * 60)
print(f"📊 ISPEZIONE SU {len(gt_files)} FRAME:")
print(f"  • Totale Ground Truth 'Car': {total_gt_cars}")
print(f"  • Totale Predizioni 'Car':   {total_pred_cars}")
print(f"  • Massima IoU 3D trovata (Standard):     {max_iou_standard:.4f}")
print(f"  • Massima IoU 3D trovata (Y-Aggiustata): {max_iou_adjusted_y:.4f}")
print("=" * 60)
```

```python
import os
import glob
import numpy as np
from shapely.geometry import Polygon
from evaluate import load_kitti_txt

exp_dir = "/content/drive/MyDrive/3D_Perception/experiments/baseline_100perc"
gt_dir = os.path.join(exp_dir, "gt_labels")
pred_dir = os.path.join(exp_dir, "pred_labels")

gt_files = sorted(glob.glob(os.path.join(gt_dir, "*.txt")))


def make_bev_poly(loc, dim, ry, swap_wl=False):
    h, w, l = dim
    if swap_wl:
        w, l = l, w
    x, y, z = loc

    x_corners = [w / 2, w / 2, -w / 2, -w / 2]
    z_corners = [l / 2, -l / 2, -l / 2, l / 2]

    R = np.array([
        [np.cos(ry), np.sin(ry)],
        [-np.sin(ry), np.cos(ry)]
    ])

    corners = np.vstack([x_corners, z_corners])
    corners = np.dot(R, corners)
    corners[0, :] += x
    corners[1, :] += z

    return Polygon(zip(corners[0, :], corners[1, :]))


def fast_iou_3d(gt_loc, gt_dim, gt_ry, p_loc, p_dim, p_ry, swap_wl, yaw_offset, y_mode):
    # 1. Aggiustamento Y
    p_loc_adj = p_loc.copy()
    if y_mode == 'center_to_bottom':
        p_loc_adj[1] += p_dim[0] / 2.0
    elif y_mode == 'bottom_to_center':
        p_loc_adj[1] -= p_dim[0] / 2.0

    # Overlap Y
    gt_y_min, gt_y_max = gt_loc[1] - gt_dim[0], gt_loc[1]
    p_y_min, p_y_max = p_loc_adj[1] - p_dim[0], p_loc_adj[1]

    inter_y = max(0.0, min(gt_y_max, p_y_max) - max(gt_y_min, p_y_min))
    if inter_y <= 0:
        return 0.0

    # 2. Overlap BEV
    try:
        p_ry_adj = p_ry + yaw_offset
        poly_gt = make_bev_poly(gt_loc, gt_dim, gt_ry, swap_wl=False)
        poly_p = make_bev_poly(p_loc_adj, p_dim, p_ry_adj, swap_wl=swap_wl)

        if not poly_gt.is_valid or not poly_p.is_valid:
            return 0.0

        inter_bev = poly_gt.intersection(poly_p).area
        if inter_bev <= 0:
            return 0.0

        vol_inter = inter_bev * inter_y
        vol_gt = poly_gt.area * gt_dim[0]
        vol_p = poly_p.area * p_dim[0]

        vol_union = vol_gt + vol_p - vol_inter
        return float(vol_inter / vol_union) if vol_union > 0 else 0.0
    except Exception:
        return 0.0


# Test Combinatorio
swap_opts = [False, True]
yaw_opts = [0.0, np.pi / 2, -np.pi / 2, np.pi]
y_opts = ['none', 'center_to_bottom', 'bottom_to_center']

best_max_iou = 0.0
best_config = None

print("🔍 Ricerca automatica trasformazione geometrica ottimale in corso...\n")

for swap_wl in swap_opts:
    for yaw_off in yaw_opts:
        for y_m in y_opts:
            max_iou = 0.0
            for gt_path in gt_files[:50]:  # Campione sui primi 50 frame
                fname = os.path.basename(gt_path)
                gt = load_kitti_txt(gt_path, is_pred=False)
                pred = load_kitti_txt(os.path.join(pred_dir, fname), is_pred=True)

                if gt is None or pred is None:
                    continue

                gt_mask = (gt['name'] == 'Car')
                p_mask = (pred['name'] == 'Car')

                for g_loc, g_dim, g_ry in zip(gt['location'][gt_mask], gt['dimensions'][gt_mask], gt['rotation_y'][gt_mask]):
                    for p_loc, p_dim, p_ry in zip(pred['location'][p_mask], pred['dimensions'][p_mask], pred['rotation_y'][p_mask]):
                        iou = fast_iou_3d(g_loc, g_dim, g_ry, p_loc, p_dim, p_ry, swap_wl, yaw_off, y_m)
                        if iou > max_iou:
                            max_iou = iou

            print(f"Config -> Swap W/L: {str(swap_wl):<5} | Yaw Shift: {np.degrees(yaw_off):>6.1f}° | Y-Mode: {y_m:<16} => Max IoU: {max_iou:.4f}")

            if max_iou > best_max_iou:
                best_max_iou = max_iou
                best_config = (swap_wl, yaw_off, y_m)

print("\n" + "=" * 60)
print(f"🏆 TRASFORMAZIONE VINCENTE TROVATA! IoU Max: {best_max_iou:.4f}")
print(f"  • Scambio Larghezza/Lunghezza (w <-> l): {best_config[0]}")
print(f"  • Rotazione Yaw (r_y): {np.degrees(best_config[1]):.1f}°")
print(f"  • Correzione Asse Y: {best_config[2]}")
print("=" * 60)
```

```python
%%writefile /content/test_calib_iou.py
import os
import sys
import numpy as np
from shapely.geometry import Polygon

# Inseriamo la repo nel path
sys.path.insert(0, "/content/3D_Perception")

from src import KittiDataset, LidarDetector

# Caricamento Dataset e Detector
data_path = "/content/drive/MyDrive/3D_Perception/data/kitti_validation"
dataset = KittiDataset(data_root=data_path)
detector = LidarDetector(conf_threshold=0.3)

# Cerchiamo il primo frame con almeno un'auto GT
sample_idx = 0
for idx in range(len(dataset)):
    gt_boxes = dataset[idx]['gt_boxes']
    if any(obj['type'] == 'Car' for obj in gt_boxes):
        sample_idx = idx
        break

sample = dataset[sample_idx]
calib = sample['calib']
detections = detector.detect(sample)

print(f"🔍 ANALISI FRAME TEST #{sample_idx}:")

def convert_lidar_to_kitti_cam(loc_lidar, dim_lidar, yaw_lidar, calib):
    """Converte un box 3D dallo spazio LiDAR a Camera Rettificato KITTI."""
    dx, dy, dz = dim_lidar
    
    # 1. Trasformazione centro 3D tramite la matrice di calibrazione (LiDAR -> Camera Rect)
    loc_cam = calib.lidar2rect(np.array([loc_lidar]))[0]
    
    # 2. Spostamento Y dal centro della scatola alla BASE INFERIORE (a contatto con l'asfalto)
    loc_cam[1] += dz / 2.0
    
    # 3. Mappatura dimensioni KITTI [h, w, l] = [dz, dy, dx]
    h, w, l = dz, dy, dx
    
    # 4. Trasformazione rotazione Yaw
    ry = -yaw_lidar - (np.pi / 2.0)
    ry = (ry + np.pi) % (2 * np.pi) - np.pi
    
    return loc_cam, [h, w, l], ry

def compute_bev_poly(loc, dim, ry):
    h, w, l = dim
    x, y, z = loc
    x_corners = [w/2, w/2, -w/2, -w/2]
    z_corners = [l/2, -l/2, -l/2, l/2]
    R = np.array([[np.cos(ry), np.sin(ry)], [-np.sin(ry), np.cos(ry)]])
    corners = np.dot(R, np.vstack([x_corners, z_corners]))
    corners[0, :] += x
    corners[1, :] += z
    return Polygon(zip(corners[0, :], corners[1, :]))

def iou_3d_test(gt_loc, gt_dim, gt_ry, p_loc, p_dim, p_ry):
    gt_y_min, gt_y_max = gt_loc[1] - gt_dim[0], gt_loc[1]
    p_y_min, p_y_max = p_loc[1] - p_dim[0], p_loc[1]
    inter_y = max(0.0, min(gt_y_max, p_y_max) - max(gt_y_min, p_y_min))
    if inter_y <= 0:
        return 0.0
    
    poly_gt = compute_bev_poly(gt_loc, gt_dim, gt_ry)
    poly_p = compute_bev_poly(p_loc, p_dim, p_ry)
    if not poly_gt.is_valid or not poly_p.is_valid:
        return 0.0
    
    inter_bev = poly_gt.intersection(poly_p).area
    if inter_bev <= 0:
        return 0.0
    
    vol_inter = inter_bev * inter_y
    vol_gt = poly_gt.area * gt_dim[0]
    vol_p = poly_p.area * p_dim[0]
    return float(vol_inter / (vol_gt + vol_p - vol_inter))

# Test IoU
gt_cars = [obj for obj in sample['gt_boxes'] if obj['type'] == 'Car']
max_found_iou = 0.0

print(f"  • Auto GT nel frame: {len(gt_cars)}")
print(f"  • Predizioni totali nel frame: {len(detections)}")

for det in detections:
    if det.type != 'Car':
        continue
    
    loc_cam, dim_cam, ry_cam = convert_lidar_to_kitti_cam(
        det.location_3d, det.dimensions_3d, det.rotation_y, calib
    )
    
    for gt in gt_cars:
        iou = iou_3d_test(gt['location_3d'], gt['dimensions_3d'], gt['rotation_y'], loc_cam, dim_cam, ry_cam)
        if iou > max_found_iou:
            max_found_iou = iou

print("\n" + "="*55)
print(f"🎯 IoU 3D MASSIMO CON CALIBRAZIONE TRASFORMATA: {max_found_iou:.4f}")
print("="*55)
```
